## Verificación del entorno

Se confirma que el notebook se está ejecutando con el **intérprete correcto**: el del
entorno virtual `.venv` del proyecto, y no el Python global del sistema. Esto es clave
para la reproducibilidad, ya que garantiza que se usen las versiones de las dependencias
instaladas en `requirements.txt`.

- **`sys.executable`** — ruta del intérprete de Python en uso.
- **`sys.version`** — versión de Python.
- **`platform.platform()`** — sistema operativo y arquitectura.
- **`assert ".venv" in sys.executable.lower()`** — falla si el kernel activo **no** es
  el del entorno virtual, avisando de inmediato que hay que seleccionar el kernel `.venv`.

```python
import sys, platform
print(sys.executable)
print(sys.version)
print(platform.platform())
assert ".venv" in sys.executable.lower()
```

### Comentario de la salida

La salida muestra la ruta del intérprete (que debe contener `.venv`), la versión de
Python y la plataforma. Que el `assert` pase sin error **confirma que el kernel activo
es el del entorno virtual del proyecto**, por lo que las dependencias y resultados son
reproducibles. Si el `assert` fallara, significa que el notebook está usando otro
Python: habría que cambiar el kernel al `.venv` (selector arriba a la derecha) y volver
a ejecutar.
